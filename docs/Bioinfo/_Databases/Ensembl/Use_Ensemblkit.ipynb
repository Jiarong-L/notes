## Seq

```text
flank:       5' flank---[region:gene/exon/etc.]---3' flank
------------------------------------------

Gene:        5'---promoter---[exon1---intron1---exon2---intron2---exon3]---3'
pre-mRNA:    5'cap---[exon1---intron1---exon2---intron2---exon3]---3'     then Alternative Splicing + 5'cap + 3'polyA
mature mRNA: 5'cap---5'UTR---CDS---3'UTR---poly(A)---3'                 then Reverse Transcription (cDNA) or Translation (peptide)
                             CDS_isoform A: Exon1-Exon2-Exon3
                             CDS_isoform B: Exon1-Exon3

peptide:     N'---Met---aa2---aa3---...---aaN---C'      [pep.fa.bgz]

cDNA:        5'---5'UTR---CDS---3'UTR---3'              for spliced mRNA (i.e. mature mRNA),  [cdna.fa.bgz]
                          CDS/ORF: ATG...stop
```

Hint: 

- download cdna.fa.bgz takes too long therefore here we can retrieve it from the genome
- peptide must be retrieved from pep.fa.bgz, since the coden might be
- still there are options to download & use cdna.fa.bgz / pep.fa.bgz


```
Get sequences, default: according to coding strand
## et.unspliced_seq  ---- unspliced
## et.transcript_seq  ---- spliced = CDS + UTR     [default: cdna.fa.bgz, missing: generate by this packege]
## et.coding_seq  ---- CDS 
## et.utr_seq  ---- UTR 
## et.protein_seq  ---- peptide     [pep.fa.bgz]
## et.flank_seq  ---- upstream/downstream
```

### Extract unspliced gene or transcript

```
Input:  id/ids for gene/transcript/exon   
Return: gene(DNA) or transcript(DNA/RNA), Default: align with the strand direction (ignore_strand = False )


Extract unspliced DNA from Genome FASTA:  gene or pre-mRNA(transcript_unspliced)
       ---  map_to="gene"/"transcript"     ===>  map the input_ids to relevant gene_ids or transcript_ids 
       ---  seq_type="dna"/"rna"           ===>  return as dna/rna sequence
       ---  seq_type="dna",ignore_strand = True  ===>  to see the original seqtext in the genome file


---------Assume the gene locates on strand = -1--------------------
Default: ignore_strand = False     
    dna  ===> Genome.extract(gene_region).ReverseComplement()
              Genome.extract(transcript_region).ReverseComplement()
    rna  ===> Genome.extract(gene_region).ReverseComplement().T2U() 
              Genome.extract(transcript_region).ReverseComplement().T2U()    i.e.pre-mRNA
Case:ignore_strand = True
    dna  ===> Genome.extract(gene_region)
              Genome.extract(transcript_region)
    rna  ===> Genome.extract(gene_region).T2U()
              Genome.extract(transcript_region).T2U()    i.e.pre-mRNA

---------Assume the gene locates on strand = 1--------------------
Case:ignore_strand = False/True
    dna  ===> Genome.extract(gene_region)
              Genome.extract(transcript_region)
    rna  ===> Genome.extract(gene_region).T2U()
              Genome.extract(transcript_region).T2U()    i.e.pre-mRNA
```

Note: transcript-gene are usually 1v1, but exon-transcript can be n-n. 

````python
## Example1: exon-transcript-gene: ENSE00004246449 - ENST00000831145 - ENSG00000310526	--- non Coding
## Example2: exon-transcript-gene: ENSE00003812156 - ENST00000641515 - ENSG00000186092	--- coding
et.unspliced_seq(ds, ["ENSE00003812156", "ENST00000641515"], map_to="gene", seq_type="dna",duplicate = False)
````

````python
et.coding_seq(ds, "ENSG00000186092")
````

````python
et.utr_seq(ds, ["ENSG00000186092","ENST00000641515"])
````

````python
def test_spliced_seqs(input_id):
    utr = et.utr_seq(ds, input_id)[0][2]
    coding = et.coding_seq(ds, input_id)[0][2]
    transcript = et.transcript_seq(ds, input_id)[0][2]
    assert utr['5_prime'] + coding +  utr['3_prime'] == transcript

test_spliced_seqs("ENST00000641515")
````
